```python
from google.colab import auth
import google.auth
import gspread

# 1. Autenticación del usuario de Google en Colab
auth.authenticate_user()

# 2. Obtención de credenciales predeterminadas
creds, _ = google.auth.default()
gc = gspread.authorize(creds)

# 3. Apertura del libro por ID
libro = gc.open_by_key(SPREADSHEET_ID)
hoja_inventario = libro.worksheet("Inventario")
hoja_reposicion = libro.worksheet("Reposicion")

# Verificación de lectura inicial
registros_prueba = hoja_inventario.get_all_records()
print(f"Conexión exitosa. Filas leídas en 'Inventario': {len(registros_prueba)}")
for r in registros_prueba[:3]:
    print("  ->", r)
```

```python
ENCABEZADOS_ESPERADOS = ["codigo", "producto", "local", "stock_actual", "stock_minimo", "stock_maximo"]

def validar_fila(fila: dict, num_fila: int) -> tuple[bool, str]:
    """
    Actividad B: Validación de reglas de integridad
    - Campos obligatorios no vacíos.
    - Valores enteros no negativos.
    - stock_maximo >= stock_minimo.
    """
    for campo in ENCABEZADOS_ESPERADOS:
        if campo not in fila or str(fila[campo]).strip() == "":
            return False, f"Fila {num_fila}: Campo obligatorio '{campo}' ausente o vacío."

    try:
        actual = int(fila["stock_actual"])
        minimo = int(fila["stock_minimo"])
        maximo = int(fila["stock_maximo"])
    except ValueError:
        return False, f"Fila {num_fila}: Los stocks deben ser números enteros."

    if actual < 0 or minimo < 0 or maximo < 0:
        return False, f"Fila {num_fila}: Los valores de stock no pueden ser negativos."

    if maximo < minimo:
        return False, f"Fila {num_fila}: 'stock_maximo' ({maximo}) no puede ser menor a 'stock_minimo' ({minimo})."

    return True, "Válido"

def obtener_ordenes_solicitadas_existentes(hoja_rep) -> set[tuple[str, str]]:
    """
    Actividad D: Lee la hoja Reposicion y extrae combinaciones (codigo, local)
    cuyo estado sea exactamente 'SOLICITADA'.
    """
    registros = hoja_rep.get_all_records()
    solicitadas = set()
    for reg in registros:
        if str(reg.get("estado", "")).strip().upper() == "SOLICITADA":
            clave = (str(reg.get("codigo", "")).strip(), str(reg.get("local", "")).strip())
            solicitadas.add(clave)
    return solicitadas

def procesar_reposicion(hoja_inv, hoja_rep):
    """
    Actividades B, C y D combinadas:
    Lee inventario, valida, calcula y añade solo si no está duplicado.
    """
    # Validar encabezados
    encabezados_reales = hoja_inv.row_values(1)
    if encabezados_reales != ENCABEZADOS_ESPERADOS:
        raise ValueError(f"Encabezados inválidos. Esperados: {ENCABEZADOS_ESPERADOS}, Encontrados: {encabezados_reales}")

    filas = hoja_inv.get_all_records()
    ordenes_existentes = obtener_ordenes_solicitadas_existentes(hoja_rep)

    # Rastrear duplicados dentro del lote actual para idempotencia estricta
    ordenes_en_este_lote = set()
    nuevas_filas_a_insertar = []
    fecha_actual = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    print("\n>>> INICIANDO ANÁLISIS DE INVENTARIO...")
    for idx, fila in enumerate(filas, start=2):
        es_valida, mensaje_error = validar_fila(fila, idx)
        if not es_valida:
            print(f"[REGLA B] Error de validación: {mensaje_error}")
            continue

        codigo = str(fila["codigo"]).strip()
        local = str(fila["local"]).strip()
        actual = int(fila["stock_actual"])
        minimo = int(fila["stock_minimo"])
        maximo = int(fila["stock_maximo"])

        # Actividad C: Comparación estricta actual < minimo
        if actual < minimo:
            clave_solicitud = (codigo, local)

            # Actividad D: Validar duplicados contra hoja y contra el lote actual
            if clave_solicitud in ordenes_existentes or clave_solicitud in ordenes_en_este_lote:
                print(f"[REGLA D - OMITIDO] Ya existe orden SOLICITADA para {codigo} en {local}.")
                continue

            cantidad_pedir = maximo - actual
            nueva_orden = [
                fecha_actual,
                codigo,
                fila["producto"],
                local,
                cantidad_pedir,
                "SOLICITADA"
            ]
            nuevas_filas_a_insertar.append(nueva_orden)
            ordenes_en_este_lote.add(clave_solicitud)
            print(f"[REGLA C - GENERADA] {codigo} ({local}): Stock {actual} < {minimo}. Cantidad a pedir: {cantidad_pedir}")
        else:
            print(f"[REGLA C - OK] {codigo} ({local}): Stock suficiente ({actual} >= {minimo}).")

    # Escritura en lote si hay órdenes nuevas
    if nuevas_filas_a_insertar:
        hoja_rep.append_rows(nuevas_filas_a_insertar)
        print(f"\nSe escribieron exitosamente {len(nuevas_filas_a_insertar)} órdenes en 'Reposicion'.")
    else:
        print("\nNo se requirieron nuevas órdenes de reposición.")

    return nuevas_filas_a_insertar
```

```python
def enviar_notificacion(ordenes_nuevas: list):
    """
    Actividad E: Notifica solo cuando hay órdenes nuevas.
    Modo prueba funciona de forma autónoma sin exigir secretos.
    """
    if not ordenes_nuevas:
        return

    resumen = f"📦 *FarmaRed - Alerta de Reposición ({len(ordenes_nuevas)} órdenes)*\n"
    for o in ordenes_nuevas:
        resumen += f"• {o[1]} - {o[2]} ({o[3]}): Pedir {o[4]} unidades [Estado: {o[5]}]\n"

    if MODO_ALERTA == 'prueba':
        print("\n" + "="*50)
        print(" [MODO PRUEBA - ALERTA SIMULADA EN CONSOLA]")
        print("="*50)
        print(resumen)
        print("="*50)
    elif MODO_ALERTA in ['discord', 'slack']:
        from google.colab import userdata
        try:
            webhook_url = userdata.get('WEBHOOK_URL')
            payload = {"content": resumen} if MODO_ALERTA == 'discord' else {"text": resumen}
            resp = requests.post(webhook_url, json=payload, timeout=10)
            if resp.status_code in [200, 204]:
                print(f"Notificación enviada con éxito a {MODO_ALERTA}.")
            else:
                print(f"Fallo al enviar notificación ({resp.status_code}): {resp.text}")
        except Exception as e:
            print(f"Error al despachar webhook: {e}")
```

```python
# Botón interactivo independiente
boton_prueba = widgets.Button(
    description="Probar notificación",
    button_style="info",
    icon="bell"
)
salida_widget = widgets.Output()

def al_presionar_boton(b):
    with salida_widget:
        salida_widget.clear_output()
        print("🔔 [TEST INTERACTIVO] Simulando alerta sin alterar Google Sheets...")
        orden_ficticia = [
            [datetime.now().strftime("%Y-%m-%d %H:%M:%S"), "MED999", "Suero Oral 500ml", "Quito Centro", 30, "SOLICITADA"]
        ]
        # Forzar modo prueba para garantizar seguridad
        modo_anterior = globals().get('MODO_ALERTA')
        globals()['MODO_ALERTA'] = 'prueba'
        enviar_notificacion(orden_ficticia)
        globals()['MODO_ALERTA'] = modo_anterior

boton_prueba.on_click(al_presionar_boton)
display(boton_prueba, salida_widget)
```

```python
# Ejecución del pipeline
ordenes_creadas = procesar_reposicion(hoja_inventario, hoja_reposicion)
enviar_notificacion(ordenes_creadas)
```
